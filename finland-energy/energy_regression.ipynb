# Finnish energy consumption regression

Compare ordinary linear regression and Lasso regression for total energy consumption in Finland.
The original experiment uses 11 rows for training and the next 3 for testing.

Place `energy_data.csv` beside this notebook and run all cells in order. See the project README for the required columns and limitations. Saved outputs are cleared; results must be regenerated from the original data.

## Load the data

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_absolute_error

# Load the dataset (Preprocess stage)
rdata = pd.read_csv('energy_data.csv')
print(rdata.head(7))
```

## Fit linear and Lasso models

```python
data = rdata.drop(columns=['Year', 'Energy Consumption Total(TJ)'])

X = np.array([[data[col][entry] for col in data.columns] for entry in range(data.shape[0])])
X_train = X[0:11]
X_test = X[11:14]

y = np.array(rdata['Energy Consumption Total(TJ)'])
y_train = y[0:11]
y_test = y[11:14]

chosen_alpha = 0.001

# Creating the models
lnregtool = LinearRegression()
lsregtool = Lasso(alpha=chosen_alpha)

# Training the model
lnreg_result = lnregtool.fit(X_train, y_train)
lsreg_result = lsregtool.fit(X_train, y_train)

# Coefficient of determination on the training set (R-squared)
print("linear training R-squared - ", lnreg_result.score(X_train, y_train))
print("lasso training R-squared - ", lsreg_result.score(X_train, y_train))

# beta1, beta2, ...
print("linear coefficients - ", lnreg_result.coef_)
print("lasso coefficients - ", lsreg_result.coef_)

# beta0
print("linear intercept - ", lnreg_result.intercept_)
print("lasso intercept - ", lsreg_result.intercept_)




```

## Compare prediction errors

```python
# Predicting

ln_preds_train = lnreg_result.predict(X_train)
ln_preds_test = lnreg_result.predict(X_test)

ls_preds_train = lsreg_result.predict(X_train)
ls_preds_test = lsreg_result.predict(X_test)

# Compare both models using mean absolute error (MAE) in target units.

print('Linear errors:')
print('\ttraining MAE - ', mean_absolute_error(y_train, ln_preds_train))
print('\ttesting MAE - ', mean_absolute_error(y_test, ln_preds_test))


print('Lasso errors:')
print('\ttraining MAE - ', mean_absolute_error(y_train, ls_preds_train))
print('\ttesting MAE - ', mean_absolute_error(y_test, ls_preds_test))
```
